# 02_标签Y
- Y列名：default payment next month，即要预测的目标（下个月是否违约）；
- 1 = 下月违约；0 = 未违约；
- 违约率 = （Y=1 的人数）÷（总人数）

```python
df["default payment next month"].value_counts()  # 查看多少人违约，多少人未违约
```

# 03_选择客户维度
选用两个维度对比违约率：
- LIMIT_BAL（授信额度）——后面分成低/中/高
- AGE（年龄）——后面分几档

```python
df["LIMIT_BAL"].describe()  # 看额度统计信息
```

```python
df["AGE"].describe()  # 看年龄统计信息
```

# 04_客群分档
用pd.cut区间切分：
- 额度：按5万、14万分成低/中/高
- 年龄：按30、40、50分为四档

```python
df["额度档"] = pd.cut(df["LIMIT_BAL"], bins=[0, 50000, 140000, 1000000], labels=["低额度", "中额度", "高额度"])
df["额度档"].value_counts()
```

```python
df["年龄档"] = pd.cut(df["AGE"], bins=[20, 30, 40, 50, 80], labels=["30岁及以下", "31-40岁", "41-50岁", "50岁以上"])
df["年龄档"].value_counts()
```

# 05_算各档违约率
- 按额度档、年龄档分别统计：人数、违约人数、违约率
- 违约率 = 违约人数 / 该档总人数

```python
整体违约率 = df["default payment next month"].mean()  # 违约是1，求平均就是在算1的占比，即整体违约率
print("整体违约率：", 整体违约率)
```

```python
额度结果 = df.groupby("额度档")["default payment next month"].aggregate(人数='count', 违约人数='sum', 违约率='mean')
额度结果
```

```python
年龄结果 = df.groupby("年龄档")["default payment next month"].aggregate(人数='count', 违约人数='sum', 违约率='mean')
年龄结果
```

# 09_SQL复核

```python
import sqlite3  # 导入Python自带的SQLite库
conn = sqlite3.connect(":memory:")  # 建立连接
df.to_sql("credit", conn, index=False, if_exists="replace")  # 将表df导出为sql表credit
print("已将表df导出为sql表credit")
```

```python
sql_额度 = """
select
    "额度档",
    count(*) as 人数,
    sum("default payment next month") as 违约人数,
    avg("default payment next month") as 违约率
from credit
group by 额度档
"""  # 把sql语句放进字符串，之后交给pd.read_sql执行
sql_额度 = pd.read_sql(sql_额度, conn)
sql_额度
```

```python
sql_年龄 = """
select
    "年龄档",
    count(*) as 人数,
    sum("default payment next month") as 违约人数,
    avg("default payment next month") as 违约率
from credit
group by 年龄档
"""  # 把sql语句放进字符串，之后交给pd.read_sql执行
sql_年龄 = pd.read_sql(sql_年龄, conn)
sql_年龄
```

## SQL复核说明
- 将含额度档、年龄档的表df写入SQLite表credit
- 用GROUP BY + COUNT/SUM/AVG 聚合各档人数、违约人数、违约率
- 结果与Python的groupby汇总一致，完成SQL复核
